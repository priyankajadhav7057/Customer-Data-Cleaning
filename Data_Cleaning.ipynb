```python
# clean col names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# check duplicates and remove it
duplicate_count = new_df.duplicated().sum()

print("\nDuplicate records:", duplicate_count)

# Remove duplicate rows
new_df = new_df.drop_duplicates()
```

```python

# Now Remove extra spaces from text columns


for column in df.select_dtypes(include="object").columns:
    new_df[column] = new_df[column].astype(str).str.strip()
```

```python
# STANDARDIZE COMMON TEXT VALUES
# Gender
if "Gender" in new_df.columns:

    new_df["Gender"] = (
        new_df["Gender"]
        .str.strip()
        .str.lower()
        .replace({
            "m": "Male",
            "male": "Male",
            "f": "Female",
            "female": "Female"
        })
    )
```

```python
# House Owner
if "House_Owner" in new_df.columns:

    new_df["House_Owner"] = (
        new_df["House_Owner"]
        .str.strip()
        .str.lower()
        .replace({
            "yes": "Yes",
            "y": "Yes",
            "no": "No",
            "n": "No"
        })
    )
```

```python
# Contact
if "Contact" in new_df.columns:

    new_df["Contact"] = (
        new_df["Contact"]
        .str.strip()
        .str.lower()
    )
```

```python
#  CONVERT NUMERIC COLUMNS
numeric_columns = [
    "Customer_Age",
    "Dependent_Count",
    "Zipcode",
    "Customer_Income",
    "Cust_Satisfaction_Score"
]

for column in numeric_columns:

    if column in new_df.columns:

        new_df[column] = pd.to_numeric(
            new_df[column],
            errors="coerce"
        )
```

```python
#  HANDLE MISSING VALUES
print("\nMissing values BEFORE cleaning:")

print(new_df.isnull().sum())


```

```python
# Numeric columns → median

numeric_columns = new_df.select_dtypes(
    include="number"
).columns

for column in numeric_columns:

    if new_df[column].isnull().sum() > 0:

        new_df[column] = new_df[column].fillna(
            new_df[column].median()
        )
```

```python
# Text columns → mode

text_columns = new_df.select_dtypes(
    include="object"
).columns

for column in text_columns:

    if new_df[column].isnull().sum() > 0:

        mode_value = new_df[column].mode()

        if len(mode_value) > 0:

            new_df[column] = new_df[column].fillna(
                mode_value[0]
            )
```

```python
# CHECK INVALID AGE VALUES
if "Customer_Age" in new_df.columns:

    new_df.loc[
        (new_df["Customer_Age"] < 18) |
        (new_df["Customer_Age"] > 100),
        "Customer_Age"
    ] = pd.NA

    new_df["Customer_Age"] = new_df["Customer_Age"].fillna(
        new_df["Customer_Age"].median()
    )

```

```python
if "Dependent_Count" in new_df.columns:

    new_df.loc[
        new_df["Dependent_Count"] < 0,
        "Dependent_Count"
    ] = new_df["Dependent_Count"].median()
```

```python
if "Cust_Satisfaction_Score" in new_df.columns:

   new_df.loc[
        ~new_df["Cust_Satisfaction_Score"].between(1, 5),
        "Cust_Satisfaction_Score"
    ] = new_df["Cust_Satisfaction_Score"].median()
```

```python
duplicates_after = new_df.duplicated().sum()

print("\nDuplicates AFTER cleaning:")
print(duplicates_after)
```
